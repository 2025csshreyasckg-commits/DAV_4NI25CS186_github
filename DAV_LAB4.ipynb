```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
df=pd.read_csv("/student_performance_dataset.csv")
print("First 5 rows of dataset:")
print(df.head())
```

```python
print("\nMissing valuesbefore handling:")
print(df.isnull().sum())
```

```python
df.loc[0:5, 'Study_Hours_per_week']=None
df.loc[10:15, 'Attendence_Rate']=None
df.loc[20:25, 'Past_Exam Scores']= None
df.loc[30:35, 'Final Exam Score'] = None

print("\nMissing values after inserting sasple NaNs:")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
df['Study_Hours_per_week'] = df['Study_Hours_per_week'].fillna(df['Study_Hours_per_Week'].mean())
```

```python
df['Attendence_Rate'] = df['Attendence_Rate'].fillna(df['Attendance_Rate'].median())
```

```python
df['Past_Exam Scores'] = df['Past_Exam Scores'].fillna(df['Past_Exam_Scores'].median())
```

```python
df['Final Exam Score'] = df['Final Exam Score'].fillna(df['Final_Exam_Score'].mean())
print("\nMissing values after handling:")
print(df.isnull().sum())
```

```python
df = pd.get_dummies(df, columns=['Parental_Education_Level'])
print("\nOne-Hot Encoded Columns:")
print(df.columns.tolist())
```

```python
le = LabelEncoder()
df['Pass_Fail_Encoded'] = le.fit_transform(df['Pass_Fail'])
print("\nLabel Encoded Result:")
print(df[['Pass_Fail', 'Pass_Fail_Encoded']].head())
```

```python
df['Gender_Encoded'] = le.fit_transform(df['Gender'])
df['Internet_Access_Encoded'] = le.fit_transform(df['Internet_Access_at_Home'])
df['Extracurricular_Activities_Encoded'] = le.fit_transform(df['Extracurricular_Activities'])
```

```python
df.to_csv("student_performance_cleaned.csv", index=False)
print("\nCleaned dataset saved to student_performance_cleaned.csv")


```
